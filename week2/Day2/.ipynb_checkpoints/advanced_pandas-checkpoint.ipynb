# Advanced Pandas

Groupby():
-used to split data into groups based on one or more columns.
-We use aggregation functions like (sum, mean, count, min and max )
-Remember: GroupBy = Split → Apply → Combine


```python
import pandas as pd
import matplotlib.pyplot as plt

data={
    'Region': ['North','East','West', 'East', 'South', 'West', 'North','South'],
    'Category': ['Technology', 'Helper','Furniture', 'Office Supplies','Furnicture','Office Supplies','Technology', 'Helper'],
    'Sales': [500, 700, 300, 400, 600, 800, 400, 800],
    'Profit': [50, 100, 30, 40, 60, 120, 70, 20]
}
df=pd.DataFrame(data)
print(df)

```

```python
df.style.hide(axis='index')
```

```python
# Region by Sales
region_sales=df.groupby('Region')['Sales'].sum()
print(region_sales)
```

```python
region_sales.plot(
    kind='bar',
    figsize=(10,7),
    color='skyblue',
    edgecolor='black'
    
)
plt.title('Sales By Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

```python
# Profit by Region
region_profit=df.groupby('Region')['Profit'].sum()
print(region_profit)
```

```python
region_profit.plot(
    kind='bar',
    figsize=(10,7),
    color='blue',
    edgecolor='green'
)
plt.title('Profit By Region')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.grid(axis='y')
plt.show()
```

```python
# Sales By Category:
category_sales=df.groupby('Category')['Sales'].sum()
print(category_sales)
```

```python
category_sales.plot(
    kind='bar',
    figsize=(10,7),
    color='hotpink',
    edgecolor='black'
)

plt.title('Sales By Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.grid(axis='y')
plt.show()
```

# merge():
- **used to combine two DataFrames based on one or more common columns or indexes.**
- **It works similarly to JOIN operations in SQL (INNER JOIN, LEFT JOIN, RIGHT JOIN, FULL OUTER JOIN).**
-  `Types of Join in Merge`:
  1) Inner Join
  2) Left Join
  3) Right Join
  4) Full Join
  5) Cross Join
  6) Left Anti Join
  7) Right Anti Join

```python
# Inner Join in merge()
import pandas as pd
import matplotlib.pyplot as plt

# Oders Data
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": ["C1", "C2", "C3", "C4"],
    "Sales": [500, 700, 300, 900]
})

# Customers Data
customers = pd.DataFrame({
    "Customer_ID": ["C1", "C2", "C3", "C5"],
    "Customer_Name": ["Abrar", "Sahil", "Imran", "Faiz"],
    "City": ["Bangalore", "Bhopal", "Delhi", "Mumbai"]
})
```

```python
print(orders)
```

```python
print(customers)
```

```python
inner_join=pd.merge(
    orders,                  # left: first DataFrame
    customers,               # right: Second DataFrame
    on= 'Customer_ID',       # Commom Column
    how='inner'              # Types of Join
)
print(inner_join)
```

```python
# Sales By Customers
plt.figure(figsize=(10,7))
plt.bar(
    inner_join['Customer_Name'],
    inner_join['Sales'],
    color='crimson',
    edgecolor='cyan'
)
plt.title('Sales By Customers')
plt.xlabel('Custmer_Name')
plt.ylabel('Sales')
#plt.grid()
plt.show()
```

```python
# Sales By City:
plt.figure(figsize=(10,7))
plt.bar(
    inner_join['City'],
    inner_join['Sales'],
    color='Cyan',
    edgecolor='grey'
)
plt.title('Sales By City')
plt.xlabel('City')
plt.ylabel('Sales')
plt.grid()
plt.show()
```

```python
# Left Join in Merge
left_join=pd.merge(
    orders,
    customers,
    on='Customer_ID',
    how='left'
)
left_join.style.hide(axis='index')
```

```python
# right Join in Merge
right_join=pd.merge(
    orders,
    customers,
    on='Customer_ID',
    how='right'
)
right_join.style.hide(axis='index')
```

# Pivat Table:
- **pivot_table() is used to summarize and analyze data by grouping rows and columns, just like the Pivot Table feature in Microsoft Excel**.
- data summarization tool
  

```python
# Create dataframe
data={
    'Region':['East', 'West', 'East', 'West', 'East'],
    'Product':['Laptop', 'Laptop', 'Mobile', 'Mobile', 'Laptop'],
    'Sales':[500, 700, 300, 400, 200],
    'Profit':[100, 150, 50, 80, 40],
    'Quantity':[2, 3, 1, 2, 1]
}
df=pd.DataFrame(data)
print(df)
```

```python
pivot=pd.pivot_table(
    df,                   # dataframe
    values='Sales',        # Column on which calculation is performed
    index='Product',      # Rows of the pivot table
    columns='',    # Columns of the pivot table
    aggfunc='sum'         # Aggregation function
)
print(pivot)
```
