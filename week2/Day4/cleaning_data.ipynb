# Target Address:

```python
def clean_address(address):

    replacements = {
        # Address abbreviations
        "Apartment": "Apt",
        "Flat": "Fl",
        "Floor": "Flr",
        "House Number": "H.No.",
        "Main Road": "Main Rd",
        "Street": "St",
        "Road": "Rd",
        "Colony": "Col",
        "Extension": "Ext",
        "Sector": "Sec",
        "Near": "Nr",

        # City standardization
        "New Delhi": "Delhi",
        "Bengaluru": "Bangalore",
        "Mumbai": "Mumbai",
        "Hyderabad": "Hyderabad",
        "Chennai": "Chennai",

        # Location terms
        "Metro Station": "Metro Stn",
        "Station": "Stn",
        "Pin Code": "PIN"
    }

    if pd.isna(address):
        return address

    for old, new in replacements.items():
        address = address.replace(old, new)

    return address


# Apply to complete address column
df["address"] = df["address"].apply(clean_address)

# Check output
df[["address"]].head(20)
```

```python
df['dob']=pd.to_datetime(df['dob'], errors='ceorce')
print(df['dob'])
```

```python
def format_date(date):

    if pd.isna(date):
        return date

    day = date.day

    if 11 <= day <= 13:
        suffix = "th"
    else:
        suffix = {1: "st", 2: "nd", 3: "rd"}.get(day % 10, "th")

    return date.strftime(f"{day}{suffix} %b, %Y")
```

```python
df["dob"] = pd.to_datetime(df["dob"], errors="coerce")

df["dob"] = df["dob"].apply(format_date)

df[["dob"]].head(10)
```
