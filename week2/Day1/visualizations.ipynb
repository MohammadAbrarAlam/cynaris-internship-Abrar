```python
# Remove duplicate records
print(df.drop_duplicates(inplace=True))
```

```python
import os

# Create charts folder
os.makedirs("charts", exist_ok=True)

print("Charts folder created successfully!")
```

```python
def save_chart(filename):
    plt.tight_layout()
    plt.savefig(
        f"charts/{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()
```

```python

# Apply Seaborn Theme
sns.set_theme(style="whitegrid", palette="Set2")

# Create Bar Chart
ax = region_sales.plot(
    kind="bar",
    figsize=(15, 6)
)

# Add Data Labels
ax.bar_label(
    ax.containers[0],
    fmt="%.0f",
    padding=3,
    fontsize=11
)

# Chart Title and Labels
plt.title("Sales by Region", fontsize=24, fontweight="bold")
plt.xlabel("Region", fontsize=14)
plt.ylabel("Total Sales", fontsize=14)

plt.xticks(rotation=0)

# Layout
plt.tight_layout()

# Save Chart
save_chart('Sales by Region')
plt.show()
```

# Bar Graph: Percentage 

```python
category_profit = df.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})

print(category_profit)
```

```python
category_profit["Profit Percentage"] = (
    category_profit["Profit"] / category_profit["Sales"]
) * 100

print(category_profit)
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    category_profit["Profit Percentage"],
    labels=category_profit.index,
    autopct="%1.2f%%",
    startangle=90,
    shadow=True
)

plt.title("Profit Percentage by Category",
          fontsize=16,
          fontweight="bold")
# Save Chart
save_chart('Profit Percentage By Category')

plt.show()
```

# Scatter Plot

```python
# Sales vs Profit (Category-wise)
sales_profit = df.groupby("Category")[["Sales", "Profit"]].mean()

print(sales_profit)
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    sales_profit["Sales"],
    sales_profit["Profit"],
    s=150,
    alpha=0.7
)

# Add category labels
for category, row in sales_profit.iterrows():
    plt.annotate(
        category,
        (row["Sales"], row["Profit"])
    )

plt.title("Average Sales vs Average Profit by Category")
plt.xlabel("Average Sales")
plt.ylabel("Average Profit")

plt.grid()
# Save Chart
save_chart("Average Sales vs Average Profit by Category")
plt.show()
```

```python

# Quantity vs Sales (Category-wise)

# Step 1: GroupBy
quantity_sales = df.groupby("Category")[["Quantity", "Sales"]].mean()

print(quantity_sales)

# Step 2: Apply Seaborn theme
sns.set_theme(style="whitegrid", palette="Set2")

# Step 3: Create Scatter Plot
plt.figure(figsize=(10, 5))

plt.scatter(
    quantity_sales["Quantity"],
    quantity_sales["Sales"],
    s=150,
    alpha=0.7
)

# Step 4: Add Category Labels
for category, row in quantity_sales.iterrows():
    plt.annotate(
        category,
        (row["Quantity"], row["Sales"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

# Step 5: Add Titles and Labels
plt.title("Average Quantity vs Average Sales by Category",
          fontsize=14, fontweight="bold")

plt.xlabel("Average Quantity")
plt.ylabel("Average Sales")

plt.tight_layout()
save_chart("Average Quantity vs Average Sales by Category")
plt.show()
```

# Box Plot 

```python
# Box Plot – Sales by Category
category_sales = df.groupby("Category")["Sales"].apply(list)

print(category_sales)
```

```python

# Apply Seaborn Theme
sns.set_theme(style="whitegrid", palette="Set2")

plt.figure(figsize=(10, 5))

plt.boxplot(
    category_sales.values,
    tick_labels=category_sales.index
)

plt.title("Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

save_chart('Sales Distribution by Category')
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```
