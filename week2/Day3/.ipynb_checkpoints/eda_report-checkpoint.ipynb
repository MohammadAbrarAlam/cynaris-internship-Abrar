# cleaning dataset

```python
# Missing Value
df.isnull()
```

```python
# Total Missing values
df.isnull().sum()
# 4 missing value
#df. isnull().sum().sum()
```

```python
# Multiple String items 

df.fillna({
        'State':'Unknown',
        'Region':'Unknown',
        'Ship Mode':'Unknown'
        
    }, inplace=True)
    

# Single string item:
df['State']=df['State'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Integer and Float misssing value fill with median
df['Quantity']=df['Quantity'].fillna(df['Quantity'].median())
print(df['Quantity'])
```

```python
df.isnull().sum()
```

```python
# Convert Datetime
df['Order Date']=pd.to_datetime(df['Order Date'])
print(df['Order Date'])
```

```python
# Extra Day in New Column
#df['Day']=df['Order Date'].dt.day
#print(df['Day'])
```

```python
# Extra Day-name in New Column
df['Day']=df['Order Date'].dt.day_name()
print(df['Day'])
```

```python
# Extra Month-name in New Column
df['Month']=df['Order Date'].dt.month_name()
print(df['Month'])
```

```python
# Extra Year in New Column
df['Year']=df['Order Date'].dt.year
print(df['Year'])
```

# 1) Day-wise:Sales|Profit|Discount:

```python
day_analysis = df.groupby("Day").agg(
    # Sales
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Count_Sales=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max"),

    # Profit
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Count_Profit=("Profit", "count"),
    Minimum_Profit=("Profit", "min"),
    Maximum_Profit=("Profit", "max"),

    # Discount
    Total_Discount=("Discount", "sum"),
    Average_Discount=("Discount", "mean"),
    Minimum_Discount=("Discount", "min"),
    Maximum_Discount=("Discount", "max")
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

day_analysis.head() 
```

# 2) Monthly-Wise: Sales|Profit|Discount
        1) Total Sales
        2) Average Sales
        3) Minimum Sale
        4) Maximum sales

```python
monthly_analysis = df.groupby("Month").agg(
    # Sales
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Count_Sales=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max"),

    # Profit
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Count_Profit=("Profit", "count"),
    Minimum_Profit=("Profit", "min"),
    Maximum_Profit=("Profit", "max"),

    # Discount
    Total_Discount=("Discount", "sum"),
    Average_Discount=("Discount", "mean"),
    Minimum_Discount=("Discount", "min"),
    Maximum_Discount=("Discount", "max")
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

monthly_analysis.sample(4)
```

# 3) Yearly-Wise:Sales| Profit| Discount
        1) Total Sales
        2) Average Sales
        3) Minimum Sale
        4) Maximum sales

```python
yearly_analysis = df.groupby("Year").agg(
    # Sales
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Count_Sales=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max"),

    # Profit
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Count_Profit=("Profit", "count"),
    Minimum_Profit=("Profit", "min"),
    Maximum_Profit=("Profit", "max"),

    # Discount
    Total_Discount=("Discount", "sum"),
    Average_Discount=("Discount", "mean"),
    Minimum_Discount=("Discount", "min"),
    Maximum_Discount=("Discount", "max")
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)


yearly_analysis.tail()
```

```python
# Delete the Unknown row
df = df.drop(df[df['Region'] == 'Unknown'].index)
print(df) 
```

```python
# Unique
print(df['Region'].unique())
```

# 4) Segment-Wise: Sales|Profit

```python
# Segment-wise Sales Analysis

seg = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Count_Sales=('Sales', 'count'),
    Minimum_Sales=('Sales', 'min'),
    Maximum_Sales=('Sales', 'max'),
    Total_Profit=('Profit','sum'),
    Minimum_Profit=('Profit', 'min'),
    Maximum_Profit=('Profit', 'max')

).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

seg.head() 
```

# 5) City-Wise: Sales|Profit|Discount

```python
city_analysis = df.groupby("City").agg(
    # Sales
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Count_Sales=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max"),

    # Profit
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Count_Profit=("Profit", "count"),
    Minimum_Profit=("Profit", "min"),
    Maximum_Profit=("Profit", "max"),

    # Discount
    Total_Discount=("Discount", "sum"),
    Average_Discount=("Discount", "mean"),
    Minimum_Discount=("Discount", "min"),
    Maximum_Discount=("Discount", "max")
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

city_analysis.sample(5) 
```

# 6) State-Wise:Sales|Profit|Discount

```python
# State-wise by 
state_analysis=df.groupby('State').agg(
    Total_Sales=("Sales","sum"),
    Average_Sales=('Sales','mean'),
    Total_Profit=('Profit','sum'),
    Average_Discount=('Discount','mean'),
    Count=('State','count'),
    Minimum_Sales=('Sales','min'),
    Maximum_Sales=("Sales","max"),
).sort_values("Total_Sales", ascending=False).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 40)
pd.set_option('display.width', None)

state_analysis.sample(6)

```

# 7) Region Wise: Profit| Sales|Discount

```python
# Region + Category + Sub-Category Analysis

region_analysis = df.groupby(
    ['Region', 'Category', 'Sub-Category']
).agg(
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Total_Profit=('Profit', 'sum'),
    Average_Profit=('Profit', 'mean'),
    Average_Discount=('Discount', 'mean'),
    Count=('Sales', 'count'),
    Minimum_Sales=('Sales','min'),
    Maximum_Sales=('Sales','max'),
    Minimum_Profit=('Profit','min'),
    Maximum_Profit=('Profit','max')
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

region_analysis.head()
```

# 8) Category-Wise: Sales|Profit| Discount

```python
# Category-wise Sales and Profit Analysis

category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean"),
    Count=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max")
).sort_values(
    by="Total_Sales",
    ascending=False
).reset_index() 

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

category_analysis.head()
```

# 8) Sub-Category-Wise:Profit|Sales|Discount

```python
# Sub-Category-wise Sales and Profit Analysis

subcategory_analysis = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean"),
    Count=("Sales", "count"),
    Minimum_Sales=("Sales", "min"),
    Maximum_Sales=("Sales", "max")
).sort_values(
    by="Total_Sales",
    ascending=False
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

subcategory_analysis.head()
```

# Region + Category + Sub-Category Analysis

```python
region_category = df.groupby(
    ['Region', 'Category', 'Sub-Category']
).agg(
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Total_Profit=('Profit', 'sum'),
    Average_Profit=('Profit', 'mean'),
    Average_Discount=('Discount', 'mean'),
    Count=('Sales', 'count')
).reset_index()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

region_category.head()
```

```python
loss_df = df[df["Profit"] < 0]
```

```python
category_loss = loss_df.groupby(
    [
        "Category", "Sub-Category", "Region",
        "City", "State", "Month", "Year", "Day"
    ]
).agg(
    # Loss
    Total_Loss=("Profit", "sum"),
    Loss_Orders=("Profit", "count"),
    Average_Loss=("Profit", "mean"),
    Minimum_Profit=("Profit", "min"),

    # Sales
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),

    # Quantity
    Total_Quantity=("Quantity", "sum"),
    
).reset_index()

category_loss["Total_Loss"] = category_loss["Total_Loss"].abs()
category_loss["Average_Loss"] = category_loss["Average_Loss"].abs()
category_loss["Minimum_Profit"] = category_loss["Minimum_Profit"].abs()

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

category_loss.head(10)
```

<h1 style="color: lime; font-size: 25px; text-align: center;"> Order Analysis</h1>

# 1) City-wise Order Analysis

```python
city_orders = df.groupby("City").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
).reset_index()

# Sort by highest orders
city_orders = city_orders.sort_values(
    by="Total_Orders",
    ascending=False
)

city_orders.head(10)
```

# 2) State-wise Order Analysis

```python
state_orders = df.groupby("State").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
).reset_index()

# Sort by highest orders
state_orders = state_orders.sort_values(
    by="Total_Orders",
    ascending=False
)
state_orders.head(5)
```

# 3) Region-wise Order Analysis

```python
region_orders = df.groupby("Region").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
).reset_index()

# Sort by highest orders
region_orders = region_orders.sort_values(
    by="Total_Orders",
    ascending=False
)
region_orders.head(5)
```

# 4) Day-wise Order Analysis

```python
day_orders = df.groupby("Day").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

day_orders = day_orders.sort_values(
    by="Total_Orders",
    ascending=False
)

day_orders.head()
```

# 5) Month-wise Order Analysis:

```python
month_orders = df.groupby("Month").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

month_orders = month_orders.sort_values(
    by="Total_Orders",
    ascending=False
)

month_orders.head()
```

# 6) Year-Wise Order Analysis:

```python
year_orders = df.groupby("Year").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

year_orders = year_orders.sort_values(
    by="Total_Orders",
    ascending=False
)

year_orders.head()
```

# Add Visual | Graph 

# 1) Save the every Plot 

```python
# Create charts folder
os.makedirs("charts", exist_ok=True)


def save_plot(filename, dpi=300):

    # Save current graph
    plt.savefig(
        f"charts/{filename}.png",
        dpi=dpi,
        bbox_inches="tight",
        facecolor="white"
    )

    print(f"Graph saved successfully: charts/{filename}.png")
```

# Line Plot 1:  Monthly Sales Trend

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
  data=monthly_melt,
    x="Month",
    y="Amount",
    hue="Metric",
    estimator="sum",
    marker="o",
    markerfacecolor="yellow",
    color="lime"
)

plt.title("Monthly Sales Trend")

save_plot("monthly_sales_trend")

plt.show()
plt.close()
```

# Plot-2: Daily Sales Trend:

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=day_melt,
    x="Day",
    y="Amount",
    hue="Metric",
    estimator="sum",
    marker="o",
    markerfacecolor="white",
    color="crimson"
)

plt.title("Daily Sales Trend")

save_plot("daily_sales_trend")
plt.grid(axis='y')

plt.show()
plt.close()
```

# Plot-3: Yearly Sales Trend

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=yearly_melt,
    x="Year",
    y="Amount",
    hue="Metric",
    estimator="sum",
    marker="o",
    markerfacecolor="white",
    color="blue"
)

plt.title("Yearly Sales Trend")

save_plot("yearly_sales_trend")

plt.show()
plt.close()
```

# Bar Plot -1: Region wise ( Total Sales | Total Profit)

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=region_melt,
    x="Region",
    y="Amount",
    hue="Metric",
    estimator="sum",
    palette=["blue", "lime"],
    edgecolor="black"
)

plt.title("Region Total Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend(title="Metric")
plt.grid(axis="y", linestyle="--", alpha=0.3)

save_plot("region_sales_profit_bar_chart")

plt.show()
plt.close()
```

# Bar Plot -2: Category wise ( Total Sales | Total Profit)

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=category_melt,
    x="Category",
    y="Amount",
    hue="Metric",
    estimator="sum",
    palette=["blue", "lime"],
    edgecolor="black"
)

plt.title("Region Total Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend(title="Metric")
plt.grid(axis="y", linestyle="--", alpha=0.3)

save_plot("region_sales_profit_bar_chart")

plt.show()
plt.close()
```

# Bar Plot 3: City

```python
plt.figure(figsize=(14, 7))

sns.barplot(
    data=city_melt,
    x="City",
    y="Amount",
    hue="Metric",
    palette=["green", "orange", "blue"],
)

plt.title("Top 5 City:  Sales, Profit and Discount Analysis")
plt.xlabel("City")
plt.ylabel("Amount")
plt.xticks(rotation=90)
plt.legend(title="Metric")
plt.tight_layout()

save_plot("city_sales_profit_discount")

plt.show()
plt.close()
```

# Relationship between Sales and Profit by Category

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    hue="Category",
    alpha=0.6,
    s=50
)

plt.title("Sales vs Profit by Category")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.legend(title="Category")
plt.tight_layout()

save_plot("scatter_sales_profit")

plt.show()
plt.close()
```
