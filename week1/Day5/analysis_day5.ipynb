fillna(): Fill missing values in string columns

```python
print('Before-Total missing value: ', df.isnull().sum().sum())
df.fillna(
    {
        "Ship Mode": "Unknown",
        "State": "Unknown",
        "Region": "Unknown",
    },
    inplace=True,
)
print('After-Total missing value: ', df.isnull().sum().sum())
```

Fill missing values in numeric columns

```python
df['Quantity']=df['Quantity'].fillna(df['Quantity'].median())
print("Total missing value: ", df.isnull().sum().sum())
```

drop_duplicates():Removing Total duplicate value

```python
print("Before Removing Total duplicate value:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
replace(): Replace incorrect values
```

```python
df["Region"] = df["Region"].replace("Wset", "West")
print(df["Region"])
```

```python
atype():Convert data types
```

```python
df["Quantity"] = df["Quantity"].astype('Int64')
print(df["Quantity"].dtypes)
```

strip(): Remove extra spaces

```python
df["City"] = df["City"].str.strip()
df["City"]
```

```python
drop(): Remove unwanted column
```

```python
df = df.drop(columns=['Row ID'])
print(df)
```

.str.upper(): Convert State column into uppercase

```python
df["State"] = df["State"].str.upper()
df["State"]
```

.str.lower(): Convert State column into lowercase

```python
df["State"] = df["State"].str.lower()
df["State"]
```

to_datatime():Convert to date format

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Order Date']
```

to_numeric(): Convert into numeric

```python
df['Sales'] = pd.to_numeric(df['Sales'])
df['Sales'].dtypes
```

rename():Rename column

```python
df = df.rename(columns={"Order ID": "order_id"})
print(df)
```

df[]: Filter rows using conditions

```python
df_filtered = df[df['Sales'] > 500]
print(df_filtered)
```

filter(): Select columns by names

```python
df_filtered = df.query(
    'Sales > 500 and Profit > 100'
)
print(df_filtered)
```

query(): Filter using expressions

```python
df_filtered = df[
    df['Region'].isin(['West', 'East'])
]
print(df_filtered)
```

isin(): Match multiple values

```python
df_filtered = df[
    df['Region'].isin(['West', 'East'])
]
print(df_filtered)
```

between(): Filter within a range

```python
df_filtered = df[df["Sales"].between(100, 500)]
print(df_filtered)
```

str.contains(): Search text

```python
df_filtered = df[
    df['City'].str.contains('New York')
]
print(df_filtered)
```
